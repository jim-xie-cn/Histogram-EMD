```python
csv_root_path = "/data/dataset/apo123/raw_csv/CTU-IoT-Malware-Capture-48-1.csv"
table = pd.read_csv(csv_root_path,chunksize=10000)
```

```python
for df in table:
    break
```

```python
df['label'].unique()
```

```python
mask = (df['label']=='Benign')|(df['label']=='benign')
df.loc[mask, 'label'] = 'benign'

mask = (df['label']!='Benign')&(df['label']!='benign')
df.loc[mask, 'label'] = 'abnormal'

mask = df['duration']=='-'
df.loc[mask, 'duration'] = 0
df.loc[mask, 'orig_bytes'] = 0
df.loc[mask, 'resp_bytes'] = 0
df.loc[mask, 'missed_bytes'] = 0
df.loc[mask, 'orig_pkts'] = 0
df.loc[mask, 'orig_ip_bytes'] = 0
df.loc[mask, 'resp_pkts'] = 0
df.loc[mask, 'resp_ip_bytes'] = 0


df[~mask].keys()
```
