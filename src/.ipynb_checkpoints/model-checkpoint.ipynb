```python
import h2o
import time,json
import itertools
import numpy as np
import pandas as pd
from common import preprocess,get_one_hot
from h2o.automl import H2OAutoML
from h2o.estimators import H2OGradientBoostingEstimator
from h2o.estimators import H2OSupportVectorMachineEstimator
from h2o.estimators import H2ODeepLearningEstimator
from h2o.estimators import H2OSupportVectorMachineEstimator
from h2o.estimators import H2OXGBoostEstimator
from h2o.estimators import H2ONaiveBayesEstimator
from h2o.estimators import H2ODeepLearningEstimator
from h2o.estimators import H2OGeneralizedLinearEstimator
from h2o.estimators import H2ORandomForestEstimator
```

```python
g_sample_file = "/data/dataset/ton/common/ton.csv"
```

# Read data and do preprocess

```python
df = pd.read_csv(g_sample_file,index_col=0)
df_all = preprocess(df)
```

# Initializa and train/valid/test dataset for H2o

```python
h2o.init(ip="localhost",port=54321)
df_h2o = h2o.H2OFrame(df_all)
df_train, df_valid, df_test = df_h2o.split_frame(ratios=[0.6,0.2], seed=1234)
x = df_train.columns
y = "label"
x.remove(y)
df_train[y] = df_train[y].asfactor()
df_valid[y] = df_valid[y].asfactor()
df_test[y] = df_test[y].asfactor()
print("Number of train, valid and test set : ", df_train.shape[0], df_valid.shape[0], df_test.shape[0])
```

# Train with customized models

```python
g_models = {
    "bayes":H2ONaiveBayesEstimator(training_frame=df_train,\
                                   validation_frame=df_valid),
    "glm":H2OGeneralizedLinearEstimator(training_frame=df_train,\
                                         validation_frame=df_valid,\
                                         nfolds = 4),
    "rf":H2ORandomForestEstimator(training_frame=df_train,\
                                         validation_frame=df_valid,\
                                         nfolds = 4),
    "svm":H2OSupportVectorMachineEstimator(training_frame=df_train,\
                                           validation_frame=df_valid),
    "gbm":H2OGradientBoostingEstimator(training_frame=df_train,\
                                       validation_frame=df_valid,\
                                       nfolds=4),
    "xgboost":H2OXGBoostEstimator(training_frame=df_train,\
                                  validation_frame=df_valid,\
                                  nfolds=4),
    "deeplearn":H2ODeepLearningEstimator(training_frame=df_train,\
                                         validation_frame=df_valid,\
                                         hidden=[200, 200],\
                                         nfolds = 4)
}
g_models_files = {}
```

# Save customized models

```python
for key in g_models:
    model = g_models[key]
    model.train(x=x, y=y)
    g_models_files[key] = h2o.save_model(model=model, path="./models/customized/%s"%key, force=True)
print(json.dumps(g_models_files,indent=4))
```

# Load and do performance check

#perf = model.model_performance(df_test)

#feature_interactions = model.feature_interaction()

#pred = model.predict(df_test)

```python
for key in g_models_files:
    model = h2o.load_model(g_models_files[key])
    perf = model.model_performance(df_test)
    display(perf)
```

# Train with auto models

```python
aml = H2OAutoML(max_models=10, \
                seed=1, \
                nfolds = 4, \
                keep_cross_validation_predictions=True)
aml.train(x=x, y=y ,training_frame=df_train,validation_frame=df_valid)
```

# Save auto models

```python
lb = aml.leaderboard
model_ids = list(aml.leaderboard['model_id'].as_data_frame().iloc[:,0])
for mid in model_ids:
    m = h2o.get_model(mid)
    h2o.save_model(model=m, path="./models/all/%s"%mid, force=True)
```
