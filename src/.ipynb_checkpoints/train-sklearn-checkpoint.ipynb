```python
# -*- coding: utf-8 -*- 
import sys,os
import numpy as np
import random
from matplotlib import pyplot as plt
%matplotlib inline
import pandas as pd
import seaborn as sns
from sklearn import preprocessing
import math
from common import preprocess,is_sample_qualified
from sklearn.svm import LinearSVC
from sklearn.svm import SVC
from sklearn.naive_bayes import GaussianNB
from numpy.random import RandomState
np.set_printoptions(suppress=True)
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.neural_network import MLPClassifier
from sklearn.neighbors import KNeighborsClassifier  
from sklearn.svm import SVC
from sklearn.gaussian_process import GaussianProcessClassifier
from sklearn.gaussian_process.kernels import RBF
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, AdaBoostClassifier, GradientBoostingClassifier,ExtraTreesClassifier, BaggingClassifier  
from sklearn.naive_bayes import GaussianNB
from sklearn.discriminant_analysis import QuadraticDiscriminantAnalysis, LinearDiscriminantAnalysis
from xgboost import XGBClassifier
sns.set(font_scale=2)
from warnings import filterwarnings
filterwarnings('ignore')
```

```python
g_classifiers = [
    ('Logistic Regression', LogisticRegression(verbose=True)), 
    ('Nearest Neighbors', KNeighborsClassifier(2)),  
    ('Linear SVM', SVC(kernel='linear', verbose=True)),
    ('RBF SVM', SVC(gamma=2, verbose=True)),
    ('Gaussian Process', GaussianProcessClassifier(1.0 * RBF(1.0))),  
    ('Decision Tree', DecisionTreeClassifier()),  
    ('Random Forest', RandomForestClassifier(verbose=True)), 
    ('AdaBoost', AdaBoostClassifier()),  
    ('Extra Trees', ExtraTreesClassifier(verbose=True)),
    ('GradientBoosting', GradientBoostingClassifier(verbose=True)),  
    ('Bagging', BaggingClassifier(verbose=True)),
    ('Naive Bayes', GaussianNB()),  
    ('QDA', QuadraticDiscriminantAnalysis()),  
    ('LDA', LinearDiscriminantAnalysis()),  
    ('MLP', MLPClassifier(verbose=True)),
    ('XGB', XGBClassifier(verbose=True)),
]
```

```python
df = pd.read_csv("/data/dataset/sample/base-0-1-train-1.csv",index_col=0)
df_all = preprocess(df)
df_all.keys()
```

```python
df_all
```

```python
df_train, df_test = train_test_split(df_all, test_size=0.2)
```

```python
raw_x = pd.DataFrame()
for key in df_train.keys():
    if not key == 'label':
        raw_x[key] = df_train[key]
raw_y = df_train['label']

test_x = pd.DataFrame()
for key in df_test.keys():
    if not key == 'label':
        test_x[key] = df_test[key]
test_y = df_test['label']
```

```python
result = []
models = {}
for name, clf in g_classifiers:
    clf.fit(raw_x, raw_y)
    score = clf.score(test_x,test_y)
    result.append((name,score))
    models[name] = clf
df = pd.DataFrame(result)
plt.xlabel('Classifier',fontsize=2)
plt.ylabel('Accuracy',fontsize=2)
sns.barplot(y=0, x=1, data=df)
```
